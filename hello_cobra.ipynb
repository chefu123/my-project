JupyterHub 内核没有外网 DNS。不要 `%pip` / `!pip`，换清华镜像也没用。

先跑诊断格。两个 host 都 FAIL 就改用：SSH 登录节点、离线 wheel，或让管理员装。

````python
import os, socket, sys

print("python:", sys.executable)
print("version:", sys.version)
for k in ["http_proxy", "https_proxy", "HTTP_PROXY", "HTTPS_PROXY"]:
    print(k, "=", os.environ.get(k))
for host in ["pypi.org", "pypi.tuna.tsinghua.edu.cn"]:
    try:
        print(host, socket.getaddrinfo(host, 443)[0][4])
    except Exception as e:
        print(host, "FAIL", type(e).__name__, e)
try:
    import cobra
    print("cobra already", cobra.__version__, cobra.__file__)
except Exception as e:
    print("cobra missing:", type(e).__name__, e)
````
