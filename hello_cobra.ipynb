不要在 notebook 里跑 `%pip install cobra`。

这个内核解析不了 `pypi.org`（`NameResolutionError`），pip 会失败。
本环境已经装好 COBRApy，直接 import 即可。

如果 `import cobra` 报 `ModuleNotFoundError`：点 **Restart Kernel** 后再跑下面一格。

```python
import cobra

print("python:", cobra.__file__)
print("version:", cobra.__version__)

model = cobra.Model("demo")
print("model:", model.id, "ok")
```
